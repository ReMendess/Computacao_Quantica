Primeiro vamos criar nosso ket0 que representa o estado de um vetor/QBIT

Podemos ver a parte imáginaria + 0.j

```python
from qiskit.quantum_info import Statevector
import numpy as np

# Criando um vetor

ket0 = Statevector([1, 0])
ket0
```

```python
# Print normal do vetor
print(ket0.data) # [1+0.j 0+0.j]
```

```python
# Se estiver em um Jupyter, podemos desenhar o vetor
ket0.draw() # por padrão, escreve o estado em ASCII
ket0.draw('latex') # desenha o vetor em LaTeX
```

```python
# Checa se o vetor é válido
print(ket0.is_valid()) # True
```

##Modificando um estado

Agora vamos criar um portão ou vetor X para multriplicar e mudar a amplitude de possibilidade quantica do nosso estado do QBIT

```python
X = np.array([[0, 1], [1, 0]])
X
```

Aqui estamos pegando nosso estado de vetor ket0 e passando como um array para manipular

```python
ket0 = np.array([1, 0])
ket0
```

Usando o np.dot estamos multiplicando o vetor ket0 pelo X e salvando em um ket1, gerando um novo vetor

```python
ket1 = np.dot(X, ket0)
ket1
```

Aqui podemos ver o resultado junto com a parte imaginaria

```python
ket1_state = Statevector(ket1)
ket1_state
```

```python
print(ket1_state.data) # [0.+0.j 1+0.j]
print(ket1_state.is_valid()) # True
```

Existem diversos portões/operações que podemos realizar X, Y, Z, Adamar


O portão Adamar é muito importante, pois ele cria a super-posição, probabilidade de ser 0 ou 1

#Medindo um Estado

```python
ket0 = np.array([1, 0])
ket_p = Statevector((1/np.sqrt(2))*np.array([1,1]))

print(ket_p) # [0.70710678+0.j 0.70710678+0.j]

# Calculando manualmente a probabilidade de medir zero
prob = np.abs(ket_p.inner(ket0))**2
print(prob) # 0.4999999999999999

# Calculando todas as probabilidades automaticamente
prob = ket_p.probabilities() # [0.5 0.5]
print(prob[0]) # 0.4999999999999999
```

#Criando e visualizando estados aleatórios

```python
import numpy as np
import random as rd
import matplotlib.pyplot as plt
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

theta = rd.uniform(0, np.pi)
phi   = rd.uniform(0, 2*np.pi)

alpha = np.cos(theta/2)
beta  = np.exp(phi*1.j)*np.sin(theta/2)

ket_psi = Statevector([alpha, beta])

plot_bloch_multivector(ket_psi, title='Estado psi')
plt.show()
```
