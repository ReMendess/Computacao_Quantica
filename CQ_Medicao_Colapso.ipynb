#Método probabilities

Método útil para visualizar as probabilidades, retornando até mesmo as de menor chance

```python
from qiskit.quantum_info import Statevector

psi = Statevector([
    (1 + 2j)/3,
    -2/3
])

probabilities = psi.probabilities()
print(probabilities) # [0.55555556 0.44444444]
```

#Método measure

O método measure realiza/simula o colapso e retorna apenas o estado considerado

rode várias vezes para simular

```python
resultado, estado_final = psi.measure()

print(f'Medição: {resultado}\nEstado após medição:')
display(estado_final.draw('latex'))
```

#Método sample_counts

Utilizado para simular n medições

```python
from qiskit.visualization import plot_histogram

resultados = psi.sample_counts(1000)

print(resultados)
# {
#     np.str_('0'): np.int64(560),
#     np.str_('1'): np.1nt64(440)
# }

print(resultados.values())
# dict_values([np.int64(560), np.int64(440)])

plot_histogram(resultados)
```
