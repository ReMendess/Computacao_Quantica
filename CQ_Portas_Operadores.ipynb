Os operadores **X, Y, Z e H** são portas lógicas quânticas (matrizes) que transformam o estado de um **qubit**. Cada uma atua de forma específica sobre a informação quântica.

---

## 🔹 Porta X (Pauli-X)

* Equivalente ao NOT clássico
* Troca os estados:

```
|0⟩ → |1⟩  
|1⟩ → |0⟩
```

👉 Interpretação: “inverte” o qubit

---

## 🔹 Porta Y (Pauli-Y)

* Também inverte o estado, mas adiciona fase quântica

```
|0⟩ → i|1⟩  
|1⟩ → -i|0⟩
```

👉 Interpretação: rotação mais complexa (inversão + fase)

---

## 🔹 Porta Z (Pauli-Z)

* Não troca |0⟩ e |1⟩
* Apenas altera a fase:

```
|0⟩ → |0⟩  
|1⟩ → -|1⟩
```

👉 Interpretação: muda o “sinal” do estado |1⟩

---

## 🔹 Porta H (Hadamard)

* Cria **superposição quântica**

```
|0⟩ → (|0⟩ + |1⟩)/√2  
|1⟩ → (|0⟩ - |1⟩)/√2
```

👉 Interpretação:

* Coloca o qubit em 0 e 1 ao mesmo tempo
* Base para paralelismo quântico

---

## 🧠 Intuição geométrica (esfera de Bloch)

Essas portas podem ser vistas como rotações:

* X → rotação no eixo X
* Y → rotação no eixo Y
* Z → rotação no eixo Z
* H → muda a base (leva de estado clássico para superposição)

---

## Resumo direto

| Porta | Função principal        |
| ----- | ----------------------- |
| X     | Inverte (0 ↔ 1)         |
| Y     | Inverte + adiciona fase |
| Z     | Mantém valor, muda fase |
| H     | Cria superposição       |

---


##Operador X

````python
from qiskit.quantum_info import Statevector, Operator

ket0 = Statevector([1,0])
ket1 = Statevector([0,1])
psi  = Statevector([(1+2j)/3, -2/3])

X = Operator([
    [0, 1],
    [1, 0]
])

display(X.draw('latex'))

display(ket0.evolve(X).draw('latex'))
display(ket1.evolve(X).draw('latex'))
display(psi.evolve(X).draw('latex'))
````

#Operador Y

````python
Y = Operator([
    [0, 1j],
    [1j, 0]
])

display(Y.draw('latex'))

display(ket0.evolve(Y).draw('latex'))
display(ket1.evolve(Y).draw('latex'))
display(psi.evolve(Y).draw('latex'))
````

#Operador Z

````python
Z = Operator([
    [1, 0],
    [0, -1]
])

display(Z.draw('latex'))

display(ket0.evolve(Z).draw('latex'))
display(ket1.evolve(Z).draw('latex'))
display(psi.evolve(Z).draw('latex'))
````

#Operador H - Colocar em superposição

````python
import numpy as np

H = Operator((1/np.sqrt(2))*np.array([
    [1, 1],
    [1, -1]
]))

display(H.draw('latex'))

display(ket0.evolve(H).draw('latex'))
display(ket1.evolve(H).draw('latex'))
display(psi.evolve(H).draw('latex'))
````
